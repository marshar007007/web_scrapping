```python
"""
Beginner web scraper: collects every book from books.toscrape.com
(a practice website made for learning scraping) and saves them to books.csv.

Install first:  pip install requests beautifulsoup4 pandas
Run:            python scrape_books.py
"""
import time
from urllib.parse import urljoin

import pandas as pd
import requests
from bs4 import BeautifulSoup

START_URL = "https://books.toscrape.com/catalogue/page-1.html"
HEADERS = {"User-Agent": "Mozilla/5.0 (student practice project)"}
DELAY_SECONDS = 1  # be polite: wait between requests so you don't overload the site

RATING_WORDS = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}


def get_page(url):
    """Step 3: download one page and return it as parsed HTML."""
    response = requests.get(url, headers=HEADERS, timeout=15)
    response.raise_for_status()   # stop with an error if the page failed (404, 500...)
    response.encoding = "utf-8"   # important later for Burmese text
    return BeautifulSoup(response.text, "html.parser")


def parse_books(soup, page_url):
    """Steps 4-5: find every book on the page and pull out its fields."""
    books = []
    for card in soup.select("article.product_pod"):
        link = card.select_one("h3 a")
        price = card.select_one("p.price_color")
        stock = card.select_one("p.instock.availability")
        rating = card.select_one("p.star-rating")
        books.append({
            "title": link["title"] if link else None,
            "price": price.get_text(strip=True) if price else None,
            "in_stock": ("In stock" in stock.get_text()) if stock else None,
            "rating": RATING_WORDS.get(rating["class"][1]) if rating else None,
            "url": urljoin(page_url, link["href"]) if link else None,
        })
    return books


def main():
    all_books = []
    url = START_URL
    while url:  # Step 6: keep following the "next" button until there is none
        print("Scraping", url)
        soup = get_page(url)
        all_books.extend(parse_books(soup, url))
        next_link = soup.select_one("li.next a")
        url = urljoin(url, next_link["href"]) if next_link else None
        time.sleep(DELAY_SECONDS)

    # Step 7: save. "utf-8-sig" makes Excel show Burmese and other scripts correctly.
    df = pd.DataFrame(all_books)
    df.to_csv("books.csv", index=False, encoding="utf-8-sig")
    print(f"Saved {len(df)} books to books.csv")


if __name__ == "__main__":
    main()
```
